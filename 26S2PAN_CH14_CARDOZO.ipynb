```python
# Ejercicio 8: Resolver el sistema por eliminación de Gauss
# Sistema:
# 3x1 - x2 + 4x3 = 2
# 17x1 + 2x2 + x3 = 14
# x1 + 12x2 - 7x3 = 54

# Definimos la matriz de coeficientes A y el vector b
A = np.array([ [3, -1, 4], 
              [17, 2, 1], 
              [1, 12, -7]], dtype=float)
b = np.array([2, 14, 54], dtype=float)

def eliminacion_gauss(A, b):
    # Armamos la matriz aumentada [A | b]
    M = np.hstack((A.copy(), b.reshape(-1, 1)))
    n = len(b)

    # Fase 1: eliminación hacia adelante
    for k in range(n - 1):
        # Pivoteo parcial para evitar divisiones por cero
        pivote = k + np.argmax(np.abs(M[k:, k]))
        if abs(M[pivote, k]) < 1e-12:
            raise ValueError("El sistema es singular o no tiene solución única.")

        # Si hace falta se intercambian filas
        if pivote != k:
            M[[k, pivote], :] = M[[pivote, k], :]

        # Se anulan los elementos debajo del pivote
        for i in range(k + 1, n):
            factor = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - factor * M[k, k:]

    # Fase 2: sustitución regresiva
    x = np.zeros(n, dtype=float)
    for i in range(n - 1, -1, -1):
        # Se resta la parte ya resuelta y se divide por el pivote
        # Se usa i+1:-1 porque la última columna es b y no forma parte de la matriz A
        x[i] = (M[i, -1] - np.dot(M[i, i + 1:-1], x[i + 1:])) / M[i, i]

    return x

# Resolución del sistema
x = eliminacion_gauss(A, b)

# Mostramos la solución
print("Solución del ejercicio 8:")
print("x =", x)
print("Comprobación A @ x =", A @ x)
```

```python
# Ejercicio 9: Resolver el mismo sistema usando Gauss-Jordan
# Sistema:
# 3x1 - x2 + 4x3 = 2
# 17x1 + 2x2 + x3 = 14
# x1 + 12x2 - 7x3 = 54

A = np.array([
    [3, -1, 4],
    [17, 2, 1],
    [1, 12, -7]
], dtype=float)

b = np.array([2, 14, 54], dtype=float)

def gauss_jordan(A, b):
    # Matriz aumentada
    M = np.hstack((A.copy(), b.reshape(-1, 1)))
    n = len(b)

    # Eliminación de Gauss-Jordan
    for k in range(n):
        # Pivoteo parcial
        pivote = k + np.argmax(np.abs(M[k:, k]))
        if abs(M[pivote, k]) < 1e-12:
            raise ValueError("El sistema es singular o no tiene solución única.")
        if pivote != k:
            M[[k, pivote], :] = M[[pivote, k], :]

        # Normalizamos la fila pivote
        M[k, :] = M[k, :] / M[k, k]

        # Hacemos cero el resto de la columna
        for i in range(n):
            if i != k:
                M[i, :] = M[i, :] - M[i, k] * M[k, :]

    return M[:, -1]

x = gauss_jordan(A, b)

print("Solución del ejercicio 9:")
print("x =", x)
print("Comprobación A @ x =", A @ x)
```

```python

# Ejercicio 10: Obtener la matriz triangular inferior L y la superior U
# A partir del sistema del problema 8:
# 3x1 - x2 + 4x3 = 2
# 17x1 + 2x2 + x3 = 14
# x1 + 12x2 - 7x3 = 54

A = np.array([
    [3, -1, 4],
    [17, 2, 1],
    [1, 12, -7]
], dtype=float)

b = np.array([2, 14, 54], dtype=float)

def descomposicion_LU(A):
    # Copiamos A en U para ir realizando la eliminación
    U = A.copy().astype(float)
    n = A.shape[0]

    # L es la matriz identidad
    L = np.eye(n, dtype=float)

    # Eliminación de Gauss para obtener L y U
    for k in range(n - 1):
        for i in range(k + 1, n):
            factor = U[i, k] / U[k, k]

            # Guardamos el multiplicador en L
            L[i, k] = factor

            # Actualizamos la fila de U
            U[i, k:] = U[i, k:] - factor * U[k, k:]

    return L, U

# Paso 1: Descomposición LU
L, U = descomposicion_LU(A)

print("Matriz L:")
print(L)
print("Matriz U:")
print(U)

# Verificación: A = L * U
print("A = L*U ?")
print(np.allclose(A, L @ U))

# Paso 2: Resolver Ly = b (sustitución progresiva)
y = np.zeros_like(b, dtype=float)
n = len(b)

for i in range(n):
    y[i] = (b[i] - np.dot(L[i, :i], y[:i])) / L[i, i]

print("Vector y:")
print(y)

# Paso 3: Resolver Ux = y (sustitución regresiva)
x = np.zeros_like(b, dtype=float)

for i in range(n - 1, -1, -1):
    x[i] = (y[i] - np.dot(U[i, i + 1:], x[i + 1:])) / U[i, i]

x1, x2, x3 = x
print("Solución del sistema:")
print("x1 =", x1)
print("x2 =", x2)
print("x3 =", x3)

# Verificación final
print("Comprobación A @ x =")
print(A @ x)
print("Vector b =")
print(b)
```

```python
# Ejercicio 11: Resolver el sistema por el método de Gauss-Seidel
# Sistema:
# 6x + y - z = 7
# x + 5y + z = 12
# x + y + 6z = 7

import numpy as np

# Definimos la matriz A y el vector b
A = np.array([
    [6, 1, -1],
    [1, 5, 1],
    [1, 1, 6]
], dtype=float)

b = np.array([7, 12, 7], dtype=float)

def gauss_seidel(A, b, x0=None, tol=1e-8, max_iter=1000):
    # Convertimos a arrays de punto flotante
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    n = len(b)

    # Inicialización del vector solución
    if x0 is None:
        x = np.zeros(n, dtype=float)
    else:
        x = np.array(x0, dtype=float)

    # Iteraciones
    for iteracion in range(max_iter):
        x_anterior = x.copy()

        # Actualizamos cada incógnita
        for i in range(n):
            # Suma de los términos de la fila excluyendo el elemento diagonal
            # A[i, :i] -> parte izquierda de la fila
            # A[i, i+1:] -> parte derecha de la fila
            suma = np.dot(A[i, :i], x[:i]) + np.dot(A[i, i+1:], x_anterior[i+1:])
            x[i] = (b[i] - suma) / A[i, i]

        # Criterio de paro
        if np.allclose(x, x_anterior, atol=tol, rtol=0):
            return x, iteracion + 1

    return x, max_iter

# Resolver el sistema
x, iteraciones = gauss_seidel(A, b, x0=[0, 0, 0], tol=1e-10, max_iter=1000)

# Mostrar resultado
print("Solución del ejercicio 11 por Gauss-Seidel:")
print("x =", x)
print("Número de iteraciones:", iteraciones)

# Verificación
print("Comprobación A @ x =")
print(A @ x)
print("Vector b =")
print(b)
```
