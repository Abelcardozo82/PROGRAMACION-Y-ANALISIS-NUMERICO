```python
import numpy as np
# Definimos la clase PerfilAlar para representar el perfil aerodinámico del ala. 
# Esta clase tiene métodos para calcular los coeficientes de sustentación y resistencia en función del ángulo de ataque.
# Los metodos de esta clase son coeficientes, que calcula los coeficientes de sustentación (cl) y resistencia (cd) 
# en función del ángulo de ataque (alpha_rad) utilizando las relaciones lineales proporcionadas por los parámetros cl_alpha y cd_alpha.
class PerfilAlar:
    def __init__(self, cl_alpha, cd_alpha):
        self.cl_alpha = cl_alpha
        self.cd_alpha = cd_alpha

    def coeficientes(self, alpha_rad):
        cl = self.cl_alpha * alpha_rad
        cd = self.cd_alpha * alpha_rad
        return cl, cd

# Definimos la clase CondicionesVuelo para representar las condiciones de vuelo, incluyendo la densidad del aire,
#  la velocidad de vuelo, la superficie alar y el ángulo de ataque. Esta clase convierte automáticamente el ángulo de ataque de grados a radianes.
class CondicionesVuelo:
    def __init__(self, rho, V_inf, S_w, alpha_deg):
        self.rho = rho
        self.V_inf = V_inf
        self.S_w = S_w
        # Conversión automática de grados a radianes
        self.alpha_rad = np.deg2rad(alpha_deg)

# Definimos la clase Aerodinamica para calcular las fuerzas aerodinámicas de sustentación y resistencia
#  en función del perfil alar y las condiciones de vuelo. Esta clase utiliza los coeficientes calculados 
# por la clase PerfilAlar y las condiciones proporcionadas por la clase CondicionesVuelo.
class Aerodinamica:
    def __init__(self, perfil, condiciones):
        self.perfil = perfil
        self.condiciones = condiciones

# Definimos el método calcular_fuerzas para calcular las fuerzas de sustentación (L) y resistencia (D) utilizando
#  los coeficientes de sustentación y resistencia obtenidos del perfil alar, así como las condiciones de vuelo proporcionadas. 
# La fórmula utilizada es L = q * S_w * cl y D = q * S_w * cd, donde q es la presión dinámica.
    def calcular_fuerzas(self):
        cl, cd = self.perfil.coeficientes(self.condiciones.alpha_rad)
        q = 0.5 * self.condiciones.rho * (self.condiciones.V_inf**2)
        L = q * self.condiciones.S_w * cl
        D = q * self.condiciones.S_w * cd
        return L, D

# Definimos la función principal para ejecutar el programa, solicitando al usuario que ingrese los valores necesarios
#  para calcular las fuerzas aerodinámicas y mostrando los resultados en la consola.
# Inputs del usuario
cl_alpha = float(input("Ingrese CLα: "))
cd_alpha = float(input("Ingrese CDα: "))
rho = float(input("Ingrese densidad del aire (kg/m3): "))
V_inf = float(input("Ingrese velocidad de vuelo (m/s): "))
S_w = float(input("Ingrese superficie alar (m2): "))
alpha_deg = float(input("Ingrese ángulo de ataque (grados): "))

# Creamos instancias de las clases PerfilAlar, CondicionesVuelo y Aerodinamica utilizando los valores ingresados 
# por el usuario.
perfil = PerfilAlar(cl_alpha, cd_alpha)
condiciones = CondicionesVuelo(rho, V_inf, S_w, alpha_deg)
aero = Aerodinamica(perfil, condiciones)
# Calculamos las fuerzas de sustentación y resistencia utilizando el método calcular_fuerzas de la clase Aerodinamica 
# y mostramos los resultados en la consola con dos decimales.
L, D = aero.calcular_fuerzas()
print(f"Sustentación L = {L:.2f} N")
print(f"Resistencia D = {D:.2f} N")

# Se realiza el input de los siguientes valores como ejemplopara el cálculo numérico
# cl_alpha = 5.7 [1/rad]
# cd_alpha = 0.6       # [1/rad]
# sw = 20.0            # [m^2]
# rho = 1.225          # [kg/m^3]
# alpha_deg = 5.0      # [grados]
# v_inf = 70.0         # [m/s]
```
