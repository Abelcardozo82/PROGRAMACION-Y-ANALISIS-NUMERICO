```python
# Ejercicio 6:
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def plot_point(self):
        import matplotlib.pyplot as plt
        plt.scatter(self.x, self.y, color="red", marker="o")
        plt.title("Gráfico del punto")
        plt.xlabel("Eje X")
        plt.ylabel("Eje Y")
        plt.grid(True)
        plt.show()
    
    def calculate_dist(self, other_x, other_y):
        return np.sqrt((self.x - other_x)**2 + (self.y - other_y)**2)
p1 = Point(3, 5)
p2 = Point(7, 1)

distancia = p1.calculate_dist(p2.x, p2.y)
print("La distancia entre los puntos es:", round(distancia, 2), "unidades de longitud.")

```

```python
# Ejercicio 7:
# What’s inheritance?
# ¿Qué es la herencia? 
# La herencia es un principio de la programación orientada a objetos que permite crear una nueva clase (subclase) basada en una clase existente (superclase).
# La subclase hereda atributos y métodos de la superclase, lo que permite reutilizar código y establecer relaciones jerárquicas entre clases.
# Esto facilita la extensión y personalización del comportamiento de las clases sin modificar la clase original.
# Ejemplo: class Perro(Animal): hereda de la clase Animal, por lo que Perro tiene los atributos y métodos de Animal, además de sus propios atributos y métodos específicos.
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre
    
    def hablar(self):
        print("Este animal hace un sonido.")

class Perro(Animal):   # Perro hereda de Animal
    def hablar(self):
        print(f"{self.nombre} dice guau cuando quiere salir!")

class Gato(Animal):    # Gato hereda de Animal
    def hablar(self):
        print(f"{self.nombre} dice miau cuando tiene hambre!")

p1 = Perro("Sura")
p2 = Gato("Berlin")

p1.hablar()   # Sura dice: guau cuando quiere salir!
p2.hablar()   # Berlin dice: miau cuando tiene hambre!
```

```python
# Ejercicio 8:
# How do we inherit from a superclass and add new methods?
# ¿Cómo heredamos de una superclase y agregamos nuevos métodos?
# Para heredar de una superclase, se define una nueva clase (subclase) y se coloca el nombre de la superclase entre paréntesis en la definición de la subclase.
# Luego, se pueden agregar nuevos métodos a la subclase definiéndolos dentro de la clase. La subclase puede usar los métodos heredados de la superclase y también puede tener sus propios métodos adicionales.
```

```python
# Ejercicio 9:
# When we inherit from a superclass, we need to replace a method with a new one; how do we do that?
# Cuando heredamos de una superclase, necesitamos reemplazar un método con uno nuevo; ¿cómo lo hacemos?
# Para reemplazar un método heredado de la superclase, simplemente definimos un método con el mismo nombre en la subclase.
# Esto sobrescribirá el método de la superclase, y cuando se llame al método desde un objeto de la subclase, se ejecutará la versión redefinida en lugar de la original.
```

```python
# Ejercicio 10:
# What is polymorphism?
# ¿Qué es el polimorfismo?
# El polimorfismo es un principio de la programación orientada a objetos que permite que un objeto de una clase pueda ser tratado como si fuera un objeto de otra clase, siempre que ambas clases estén relacionadas por herencia.
# Esto permite escribir código más flexible y reutilizable, ya que se puede utilizar una única interfaz para interactuar con diferentes tipos de objetos.
```

```python
# Ejercicio 11:
class Moto:
    def __init__(self, marca, modelo, color):
        self.marca = marca
        self.modelo = modelo
        self.color = color
    
    def encender(self):
        print(f"La moto {self.marca} {self.modelo} está encendida.")
    
    def apagar(self):
        print(f"La moto {self.marca} {self.modelo} está apagada.")

class MotoDeportiva(Moto):
    def __init__(self, marca, modelo, color, velocidad_max):
        super().__init__(marca, modelo, color)   # llamamos al constructor de la clase padre
        self.velocidad_max = velocidad_max
    
    def encender(self):
        print(f"La moto deportiva {self.marca} {self.modelo} ruge al encenderse.")
    
    def acelerar(self, velocidad):
        if velocidad <= self.velocidad_max:
            print(f"La moto acelera a {velocidad} km/h.")
        else:
            print("¡Velocidad excedida!")

class MotoScooter(Moto):
    def __init__(self, marca, modelo, color, autonomia):
        super().__init__(marca, modelo, color)
        self.autonomia = autonomia
    
    def encender(self):
        print(f"La scooter {self.marca} {self.modelo} se enciende suavemente.")
    
    def cargar_bateria(self):
        print(f"La scooter {self.marca} {self.modelo} está cargando su batería.")
m1 = MotoDeportiva("Yamaha", "R1", "Rojo", 300)
m2 = MotoScooter("Honda", "PCX", "Azul", 150)

m1.encender()       # La moto deportiva Yamaha R1 ruge al encenderse.
m1.acelerar(280)    # La moto acelera a 280 km/h.

m2.encender()       # La scooter Honda PCX se enciende suavemente.
m2.cargar_bateria() # La scooter Honda PCX está cargando su batería.
```
