```python
import numpy as np

# Ejercicio 1: Implementación del método de Newton-Raphson y bisección para encontrar raíces de funciones.
# Creamos una clase llamada Root_Finding que contiene dos métodos: newton_raphson y biseccion.

class Root_Finding:
    # Creamos el método newton_raphson que implementa el método de Newton-Raphson para encontrar raíces de funciones.
    def newton_raphson(self, f, df, x0, tol=1e-6, max_iter=100):

        x = x0

        for i in range(max_iter):

            if abs(f(x)) < tol:
                return x

            x = x - f(x) / df(x)

        return x
    # Creamos el método biseccion que implementa el método de bisección para encontrar raíces de funciones.
    def biseccion(self, f, a, b, tol=1e-6, max_iter=100):

        if f(a) * f(b) >= 0:
            raise ValueError("La función no cambia de signo en el intervalo.")

        for i in range(max_iter):

            c = (a + b) / 2.0

            if abs(f(c)) < tol or (b - a) / 2 < tol:
                return c

            if f(a) * f(c) < 0:
                b = c
            else:
                a = c
        return c
# ==========================================   
# MÉTODO DE BISECCIÓN
# ==========================================
# Definimos la función biseccion que implementa el método de bisección para encontrar raíces de funciones.
def biseccion(f, a, b, tol=1e-8, max_iter=100):

    if f(a) * f(b) >= 0:
        raise ValueError("La función no cambia de signo en el intervalo.")        

    for n in range(max_iter):

        c = (a + b) / 2.0

        if np.abs(f(c)) < tol or (b - a) / 2 < tol:
            return c, n + 1

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return c, max_iter

# ==========================================
# MÉTODO DE NEWTON-RAPHSON
# ==========================================
# Definimos la función Newton-Raphson que implementa el método de Newton-Raphson para encontrar raíces de funciones.
def newton(f, df, x0, tol=1e-8, max_iter=100):

    x = x0

    for n in range(max_iter):

        dfx = df(x)

        if np.abs(dfx) < 1e-12:
            raise ValueError("Derivada demasiado cercana a cero."    )

        x_new = x - f(x) / dfx

        if np.abs(x_new - x) < tol:
            return x_new, n + 1

        x = x_new

    return x, max_iter
# Planteamos un ejemplo para probar los métodos implementados. Definimos la función f(x) = x³ - x - 2 y su derivada
#  df(x) = 3x² - 1.
# ==========================================
# EJEMPLO
# f(x) = x³ - x - 2
# ==========================================
f = lambda x: x**3 - x - 2
df = lambda x: 3*x**2 - 1


# Bisección
raiz_bis, iter_bis = biseccion(f, 1.0, 2.0)

# Newton
raiz_new, iter_new = newton(f, df, 1.5)


# ==========================================
# RESULTADOS
# ==========================================
print("=== BISECCIÓN ===")
print(f"Raíz        = {raiz_bis:.10f}")
print(f"Iteraciones = {iter_bis}")

print("\n=== NEWTON-RAPHSON ===")
print(f"Raíz        = {raiz_new:.10f}")
print(f"Iteraciones = {iter_new}")
```
