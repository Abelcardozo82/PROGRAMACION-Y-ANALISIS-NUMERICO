```python
print("Ejercicio 11")
from math import factorial, sqrt

def ramanujan_pi(N):
    suma = 0
    for k in range(N + 1):
        termino = (factorial(4*k) * (1103 + 26390*k)) / (factorial(k)**4 * 396**(4*k))
        suma += termino
    return 1 / ((2 * sqrt(2) / 9801) * suma)

print("para k=0, pi es igual a", round(ramanujan_pi(0), 15))
print("para k=1, pi es igual a", round(ramanujan_pi(1), 15))
```

```python
print("Ejercicio 18")
# Verificación de las leyes de De Morgan

valores = [True, False]

print("Primera ley: ¬(P or Q) == (¬P and ¬Q)")
for P in valores:
    for Q in valores:
        izquierda = not (P or Q)
        derecha = (not P) and (not Q)
        print(f"P={P}, Q={Q} -> ¬(P or Q)={izquierda}, (¬P and ¬Q)={derecha}")

print("\nSegunda ley: ¬(P and Q) == (¬P or ¬Q)")
for P in valores:
    for Q in valores:
        izquierda = not (P and Q)
        derecha = (not P) or (not Q)
        print(f"P={P}, Q={Q} -> ¬(P and Q)={izquierda}, (¬P or ¬Q)={derecha}")


```

```python
print("Ejercicio 19")
# Tabla de verdad para (P and Q) or (P and not Q)

p_values = [True, False]
q_values = [True, False]

print("P   Q   (P and Q) or (P and not Q)")
for P in p_values:
    for Q in q_values:
        resultado = (P and Q) or (P and not Q)
        print(P, Q, resultado)
```

```python
print("Ejercicio 20")
# Pide negar la primer ley de De Morgan, que es ¬(P or Q) == (¬P and ¬Q)
p_values = [True, False]
q_values = [True, False]

print("P    Q    P or Q   ¬(¬P and ¬Q)")
for P in p_values:
    for Q in q_values:
        original = P or Q
        equivalente = not (not P and not Q)
        print(P, Q, original, equivalente)
```

```python
print("Ejercicio 21")
# Pide negar la segunda ley de De Morgan, que es ¬(P and Q) == (¬P or ¬Q)
p_values = [True, False]
q_values = [True, False]

print("P   Q   P and Q   ¬(¬P or ¬Q)")
for P in p_values:
    for Q in q_values:
        original = P and Q
        equivalente = not (not P or not Q)
        print(P, Q, original, equivalente)

```

```python
print("Ejercicio 22")
# verificación de la equivalencia entre XOR y (P or Q) and not (P and Q)
p_values = [True, False]
q_values = [True, False]

print("P   Q   XOR   (P or Q) and not (P and Q)")
for P in p_values:
    for Q in q_values:
        xor_equiv = (P or Q) and not (P and Q)
        xor_direct = P ^ Q  # operador XOR en Python
        print(P, Q, xor_direct, xor_equiv)
```
